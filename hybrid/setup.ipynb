```python
import pandas as pd
import os

csv_files = [f for f in os.listdir() if f.endswith('.csv')]

for file in csv_files:
    df = pd.read_csv(file)
    print(f"{file}: {len(df)} rows")
```

```python
def filter_by_label_and_size(csv_filename, label_value, num_rows):
    """
    Filter a CSV file to keep only specified number of rows with a given label.
    Rows with different labels are preserved unchanged.
    
    Parameters:
    - csv_filename: name of the CSV file
    - label_value: the label value to filter by
    - num_rows: number of rows to keep for that label
    """
    df_file = pd.read_csv(csv_filename)
    
    # Filter rows with the specified label
    filtered_df = df_file[df_file['label'] == label_value].head(num_rows)
    
    # Keep rows with different labels
    other_rows = df_file[df_file['label'] != label_value]
    
    # Combine filtered rows with other rows
    result_df = pd.concat([filtered_df, other_rows], ignore_index=False)
    
    # Overwrite the original CSV file
    result_df.to_csv(csv_filename, index=False)
    
    print(f"Kept {len(filtered_df)} rows with label {label_value} in {csv_filename}")
    return result_df
```

```python
# First, check if csv_files is empty and repopulate if needed
if not csv_files:
    csv_files = [f for f in os.listdir('./csvs') if f.endswith('.csv')]
    print(f"Found CSV files: {csv_files}")

if csv_files:
    # Create the main folder and subfolders
    os.makedirs('Text_Files/train', exist_ok=True)
    os.makedirs('Text_Files/test', exist_ok=True)
    os.makedirs('Text_Files/val', exist_ok=True)

    # Create label subfolders for each split
    for split in ['train', 'test', 'val']:
        for label in [0, 1]:
            os.makedirs(f'Text_Files/{split}/label_{label}', exist_ok=True)

    # Process each CSV file
    file_counter = 0
    for csv_file in csv_files:
        df = pd.read_csv(f'./csvs/{csv_file}')
        
        # Determine the subfolder based on the filename
        if 'train' in csv_file:
            split = 'train'
        elif 'test' in csv_file:
            split = 'test'
        elif 'val' in csv_file:
            split = 'val'
        else:
            continue
        
        # For each row, create a .txt file with unique counter as filename
        for idx, row in df.iterrows():
            code = row['code']
            label = row['label']
            
            filepath = os.path.join(f'Text_Files/{split}/label_{label}', f'{file_counter}.txt')
            with open(filepath, 'w') as f:
                f.write(code)
            
            file_counter += 1

    print("Text files created successfully!")
else:
    print("No CSV files found in the ./csvs directory")
```

```python
# Calculate total files in each label subfolder for each split
for split in ['train', 'test', 'val']:
    total_files = 0
    for label in [0, 1]:
        label_folder = f'Text_Files/{split}/label_{label}'
        num_files = len(os.listdir(label_folder))
        print(f"{label_folder}: {num_files} files")
        total_files += num_files
    print(f"Total files in Text_Files/{split}: {total_files}\n")
```
